```python
# Import pandas for tabular data processing.
import pandas as pd

# Import requests for sending HTTP requests to the API.
import requests

# Import os for reading environment variables.
import os

# Import load_dotenv to load variables from the .env file.
from dotenv import load_dotenv
```

```python
# Load variables stored inside the .env file.
load_dotenv()

# Read the Aviationstack API key from the environment.
API_KEY = os.getenv("AVIATIONSTACK_API_KEY")

# Check whether the API key was successfully loaded.
print("API key loaded:", API_KEY is not None)
```

```python
# Define the Aviationstack flights endpoint.
url = "https://api.aviationstack.com/v1/flights"

# Create parameters required by the API.
params = {
    "access_key": API_KEY,
    "limit": 5
}

# Send a GET request to the Aviationstack API.
response = requests.get(url, params=params)

# Print the HTTP status code.
print("Status Code:", response.status_code)

# Convert the API response into Python dictionary format.
data = response.json()

# Display the returned data.
data
```

```python
# Extract the list of flights from the API response.
flights = data["data"]

# Convert nested JSON flight information into a flat DataFrame.
api_df = pd.json_normalize(flights)

# Display the shape of the DataFrame.
print("Shape:", api_df.shape)

# Display the first few records.
api_df.head()
```
